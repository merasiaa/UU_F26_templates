```python
# n = 40 coins with 0.5 probability
# flip a 'coin' 40 times and keep count of how many are heads. tails= 0 and heads = 1

import numpy as np
from math import sqrt
n=40 #each sample has 40 coin tosses, going to need a nested loop
b= 5000 #samples
p = 0.5
rng = np.random.default_rng()

lst = []

for i in range(b):
    tailscount = 0
    headscount = 0
    for j in range(n):
        x= rng.integers(0,2)
        if x == 0:
            tailscount += 1
        if x == 1:
            headscount += 1



    heads_final = headscount - n*p

    calculated = heads_final / np.sqrt( n* p * (1-p))
    lst.append(calculated)

len(lst)

#need to sort the observation list from small to large

sortedlst = np.sort(lst)

x= sortedlst

#i needed to divide by b because im plotting proportions of observations not just the observations
y= np.arange(1,(b+1))/b

#plots
import matplotlib.pyplot as plt

plt.plot(x,y)
plt.xlabel("Heads")
plt.ylabel('proportions')
plt.title('ECDF for Num of Heads')

plt.show

```

```python
b= 5000
k=50
lst = []
rng = np.random.default_rng()


values = np.arange(1,(k+1))/k
for i in range (b):
    z = rng.choice(values)
    lst.append(z)


sortedlst = np.sort(lst)

x = sortedlst
y = np.arange(1, b + 1) / b

import matplotlib.pyplot as plt

plt.plot(x,y)
plt.xlabel("k values")
plt.ylabel('proportions')
plt.title('ECDF for Draws')

plt.show
```

```python
K= 100000
n= 5000

ranks = np.arange(1, K+1)
lst =[]
s=1
for i in range(n):
    pick = rng.choice(ranks)
    sizes = (K/pick) ** s
    lst.append(sizes)
len(lst)


sortedlst= np.sort(lst)

x= sortedlst
y= np.arange(1, n + 1) / n

plt.plot(x,y)
plt.xlabel(sizes)
plt.ylabel('proportions')
```
