```python
import pandas as pd
import matplotlib.pyplot as plt
data = [1,1.75,2,3,3.25,4]

pd.Series(data).plot.kde(bw_method = 0.2)
plt.show()
```

```python
pd.Series(data).plot.kde(bw_method = 1.42)
plt.show()
```
