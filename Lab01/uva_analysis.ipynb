```python
import pandas as pd

#1 load the foreign gifts data
df = pd.read_csv('ForeignGifts_edu.csv')
df
```

```python
df[df['Institution Name']=='University of Virginia'].groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=False)
```

```python
df[df['Institution Name']=='University of Virginia'].groupby('Country of Giftor')['Foreign Gift Amount'].sum().sort_values(ascending=True)
```

Top Giftor(cummulative sum)for UVA is Bangladesh, followed by Sweden Tanzania and Switzerland

Lowest Giftors(cummulative) are France, Japan, Rwanda, Israel

```python
df[df['Institution Name']=='University of Virginia'].groupby('Country of Giftor')['Foreign Gift Amount'].count().sort_values(ascending=False)
```

We have received the most frequent gifts from england, followed by israel and bangladesh.

Israel gifts frequently but is also among the lowest giftors when it comes to amount.
